```python
import cv2
import os
import glob

interval_sec = 0.03
root_dir = '../data'
output_root = 'sliced_images'
os.makedirs(output_root, exist_ok=True)

mp4_files = glob.glob(os.path.join(root_dir, '**', '*.mp4'), recursive=True)
print(f'총 {len(mp4_files)}개의 동영상 발견.')

for idx, video_path in enumerate(mp4_files):
    print(f'\n[{idx+1}/{len(mp4_files)}] 처리 중: {video_path}')

    #경로 분해
    parts = video_path.replace('\\', '/').split('/')

    try:
        level = parts[-2]
        pose_name = parts[-3]  #자세 이름
        actor = os.path.splitext(os.path.basename(video_path))[0]  #actorP072 등

        output_dir = os.path.join(output_root, pose_name, level, actor)
        os.makedirs(output_dir, exist_ok=True)
    except IndexError:
        print(f"경로 구조 이상: {video_path}")
        continue

    cap = cv2.VideoCapture(video_path)
    if not cap.isOpened():
        print(f"열 수 없음: {video_path}")
        continue

    fps = cap.get(cv2.CAP_PROP_FPS)
    interval_frame = max(1, int(round(fps * interval_sec)))
    total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))
    print(f"   FPS: {fps:.2f}, 총 프레임: {total_frames}, 간격: {interval_frame}프레임마다 저장")

    frame_idx = 0
    saved_idx = 0

    while cap.isOpened():
        ret, frame = cap.read()
        if not ret:
            break

        if frame_idx % interval_frame == 0:
            frame_filename = os.path.join(output_dir, f'frame_{saved_idx:06d}.jpg')
            success = cv2.imwrite(frame_filename, frame)
            if success:
                print(f"   저장됨: {frame_filename}")
            else:
                print(f" 저장 실패: {frame_filename}")
            saved_idx += 1

        frame_idx += 1

    cap.release()
    print(f"    총 저장된 이미지 수: {saved_idx}")
```
